```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x) + np.random.randn(100) * 0.2

# 颜色映射 —— 让颜色随数据值变化，适合散点图、热力图
fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(x, y, c=y, cmap='viridis', s=50, alpha=0.7)
# c=y 表示颜色由 y 值决定，cmap 指定映射方案

ax.set_title('颜色映射示例')
plt.colorbar(scatter, label='Y值')  # 添加颜色条
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)
y4 = np.cos(x) * 0.5

fig, ax = plt.subplots(figsize=(10, 6))

# 线条样式
ax.plot(x, y1, linestyle='-', linewidth=2, label='实线 (默认)')      # 实线（默认）
ax.plot(x, y2, linestyle='--', linewidth=2, label='虚线 (--)')     # 虚线
ax.plot(x, y3, linestyle='-.', linewidth=2, label='点划线 (-.)')   # 点划线
ax.plot(x, y4, linestyle=':', linewidth=2, label='点线 (:)')       # 点线

ax.set_title('不同线条样式')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 20)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * 0.5

fig, ax = plt.subplots(figsize=(10, 6))

# 标记样式 —— 控制数据点的外观
# marker: 标记形状, markersize: 标记大小, markerfacecolor: 标记填充色, markeredgecolor: 标记边框色
ax.plot(x, y1, marker='o', markersize=8, markerfacecolor='red',
        markeredgecolor='black', label='圆形 (o)', linewidth=1.5)
ax.plot(x, y2, marker='s', markersize=8, markerfacecolor='blue',
        markeredgecolor='black', label='方形 (s)', linewidth=1.5)
ax.plot(x, y3, marker='^', markersize=8, markerfacecolor='green',
        markeredgecolor='black', label='上三角 (^)', linewidth=1.5)

ax.set_title('不同标记样式')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()

# 常用标记类型: 'o'(圆), 's'(方), '^'(上三角), 'd'(菱形), '*'(星), '+'(加号)
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'SimSun', 'KaiTi']
plt.rcParams['axes.unicode_minus'] = False

fig, ax = plt.subplots()

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

ax.plot(x, y1, label='Series 1')
ax.plot(x, y2, label='Series 2')
ax.plot(x, y3, label='Series 3')

# 图例 —— 解释图中各元素的含义
# loc: 图例位置, 可选 'best'(自动), 'upper right'(右上), 'upper left'(左上),
#      'lower left'(左下), 'lower right'(右下) 等
# frameon: 是否显示图例框, shadow: 是否添加阴影
ax.legend(loc='upper right', frameon=True, shadow=True)

# 标题
fig.suptitle('Main Title', fontsize=16, fontweight='bold')   # 画布级总标题（最上方）
ax.set_title('Subtitle', fontsize=14, fontweight='bold')     # 坐标系级子标题

ax.set_xlabel('X Label', fontsize=12)
ax.set_ylabel('Y Label', fontsize=12)

# 网格 —— 让图表更易读
ax.grid(True, linestyle='--', alpha=0.7)  # 同时显示 x 和 y 轴网格
ax.grid(True, axis='x')                   # 只显示 x 轴方向的网格线

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax1 = plt.subplots()

x = np.linspace(0, 10, 100)
y1 = np.sin(x)               # 第一组数据：正弦曲线
y2 = np.exp(x)               # 第二组数据：指数曲线（数量级差异大）

# 在 ax1（左轴）上绘制 sin(x)
ax1.set_xlabel('X')                          # x 轴标签
ax1.set_ylabel('sin(X)', color='blue')       # 左 y 轴标签，蓝色
ax1.plot(x, y1, color='blue')                # 绘制蓝色正弦曲线
ax1.tick_params(axis='y', labelcolor='blue') # 将左 y 轴刻度颜色也设为蓝色

ax2 = ax1.twinx()
ax2.set_ylabel('exp(X)', color='red')        # 右 y 轴标签，红色
ax2.plot(x, y2, color='red')                 # 绘制红色指数曲线
ax2.tick_params(axis='y', labelcolor='red')  # 将右 y 轴刻度颜色也设为红色

plt.show()
```

```python
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import axes3d

ax = plt.figure().add_subplot(projection='3d')
X, Y, Z = axes3d.get_test_data(0.05)

# Plot the 3D surface
ax.plot_surface(X, Y, Z, edgecolor='royalblue', lw=0.5, rstride=8, cstride=8,
                alpha=0.3)

# Plot projections of the contours for each dimension.  By choosing offsets
# that match the appropriate axes limits, the projected contours will sit on
# the 'walls' of the graph
ax.contourf(X, Y, Z, zdir='z', offset=-100, cmap='coolwarm')
ax.contourf(X, Y, Z, zdir='x', offset=-40, cmap='coolwarm')
ax.contourf(X, Y, Z, zdir='y', offset=40, cmap='coolwarm')

ax.set(xlim=(-40, 40), ylim=(-40, 40), zlim=(-100, 100),
       xlabel='X', ylabel='Y', zlabel='Z')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation  # 导入动画模块

fig, ax = plt.subplots()
x = np.linspace(0, 2 * np.pi, 100)
line, = ax.plot(x, np.sin(x))       # 绘制初始正弦曲线，line 是曲线的引用对象

def animate(frame):
    # x + frame/10：在自变量上加正数，根据"左加"原理，图像向左移动
    line.set_ydata(np.sin(x + frame / 10))  # set_ydata 只更新 y 数据，不重绘整张图
    return line,

# fig: 画布对象
# animate: 每帧调用的更新函数
# frames=100: 总共 100 帧
# interval=50: 每帧间隔 50 毫秒（即 20 fps）
# blit=True: 只重绘变化的元素，提高渲染效率
ani = FuncAnimation(fig, animate, frames=100, interval=50, blit=True)
plt.show()

# 保存动画为 GIF 文件
ani.save('animation.gif', writer='pillow')
```
